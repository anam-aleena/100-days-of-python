```python
import pandas as pd
import numpy as np 
import matplotlib.pyplot as plt
import seaborn as sns


import warnings
warnings.filterwarnings('ignore')
```

```python
df = pd.read_csv("/kaggle/input/datasets/vivek468/superstore-dataset-final/Sample - Superstore.csv", encoding='latin1')
df
```

```python
df.info()
```

```python
df.describe()
```

```python
df.describe(include='O')
```

```python
df.isna().sum()
```

```python
df['Order Date'] = pd.to_datetime(df['Order Date'])
df['Ship Date'] = pd.to_datetime(df['Ship Date'])
```

```python
df['Year'] = df['Order Date'].dt.year
df['Month'] = df['Order Date'].dt.month
df['Month Name'] = df['Order Date'].dt.strftime('%B')
```

```python
df['Shipping Days'] = (df['Ship Date'] - df['Order Date']).dt.days
```

```python
df['Profit Margin'] = df['Profit'] / df['Sales']
```

```python
aov = df.groupby('Order ID')['Sales'].sum().mean()
```

```python
sns.histplot(df['Sales'], bins=50)
plt.show()
```

```python
sns.histplot(df['Profit'], bins=50)
plt.show()
```

```python
sns.boxplot(x=df['Sales'])
plt.show()
```

```python
corr = df[['Sales','Profit','Discount','Quantity']].corr()

sns.heatmap(corr, annot=True, cmap='coolwarm')
plt.show()
```

```python
sns.scatterplot(x='Discount', y='Profit', data=df)
plt.show()
```

```python
df.groupby('Discount')['Profit'].mean()
```

```python
state_analysis = df.groupby('State').agg({
    'Sales':'sum',
    'Profit':'sum'
})

state_analysis['Profit Margin'] = state_analysis['Profit'] / state_analysis['Sales']
state_analysis.sort_values('Profit Margin')
```

```python
rfm = df.groupby('Customer ID').agg({
    'Order Date': lambda x: (df['Order Date'].max() - x.max()).days,
    'Order ID': 'nunique',
    'Sales': 'sum'
})

rfm.columns = ['Recency','Frequency','Monetary']
rfm.head()
```

```python
product_perf = df.groupby('Product Name').agg({
    'Sales':'sum',
    'Profit':'sum'
})

product_perf.sort_values('Profit').head(10) 
product_perf.sort_values('Profit', ascending=False).head(10) 
```

```python
monthly_sales = df.groupby(pd.Grouper(key='Order Date', freq='M'))['Sales'].sum()

monthly_sales.plot()
plt.show()
```

```python
df.groupby('Ship Mode')['Shipping Days'].mean()
```

```python
loss_df = df[df['Profit'] < 0]

loss_df.groupby('Sub-Category')['Profit'].sum().sort_values()
```

```python
df[df['Discount'] > 0.3]['Profit'].mean()
```

```python
df['Order Month'] = df['Order Date'].dt.to_period('M')
df['Cohort'] = df.groupby('Customer ID')['Order Date'].transform('min').dt.to_period('M')

cohort_data = df.groupby(['Cohort','Order Month'])['Customer ID'].nunique().reset_index()

cohort_pivot = cohort_data.pivot(index='Cohort', columns='Order Month', values='Customer ID')
cohort_pivot
```

```python
df.groupby('Customer ID')['Sales'].sum().mean()
```

```python
df.groupby('Order ID')['Profit'].sum().mean()
```

```python
df['Customer Type'] = pd.qcut(
    df.groupby('Customer ID')['Profit'].transform('sum'),
    q=3,
    labels=['Low Value','Medium Value','High Value']
)
```

```python
df.groupby('Customer Type').agg({
    'Sales': 'sum',
    'Profit': 'sum',
    'Customer ID': 'nunique'
})
```

```python
customer_sales = df.groupby('Customer ID')['Sales'].sum().sort_values(ascending=False)

cum_perc = customer_sales.cumsum() / customer_sales.sum()

customer_sales[cum_perc <= 0.8]
```

```python
from scipy.stats import ttest_ind

low_discount = df[df['Discount'] < 0.2]['Profit']
high_discount = df[df['Discount'] > 0.2]['Profit']

ttest_ind(low_discount, high_discount)
```

```python
loss = df[df['Profit'] < 0]

loss.groupby(['Category','Sub-Category'])['Profit'].sum()
```

```python
clv = df.groupby('Customer ID').agg({
    'Sales':'sum',
    'Order ID':'nunique'
})

clv['CLV'] = clv['Sales'] / clv['Order ID']
clv.sort_values('CLV', ascending=False)
```
